```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LogisticRegression
np.random.seed(1)
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
X = iris.data
y = iris.target
```

```python
#%pip install mlxtend
```

```python
X = pd.DataFrame(iris.data, columns=['sepal length', 'sepal width', 'petal length', 'petal width'])
```

```python
y
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
# Split data into training and test sets
X_train, X_test , y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1, stratify=y)
```

```python
y_test
```

```python
from mlxtend.feature_selection import ExhaustiveFeatureSelector as EFS
```

```python
EFS()
```

```python
from mlxtend.feature_selection import ExhaustiveFeatureSelector as EFS
# Create Logistic Regression classifier
logmodel = LogisticRegression(max_iter=1000,penalty=None)
efs = EFS(logmodel, 
           min_features=1,
           max_features=4,
           scoring='accuracy',
           print_progress=True,
           cv=5)

efs = efs.fit(X_train, y_train)

print('Best accuracy score: %.2f' % efs.best_score_)
print('Best subset (indices):', efs.best_idx_)
print('Best subset (corresponding names):', efs.best_feature_names_)
```

```python
#Via the subsets_ attribute, we can take a look at the selected feature indices at each step
efs.subsets_
```

```python
pd.DataFrame.from_dict(efs.get_metric_dict()).T
```

```python
logmodel=LogisticRegression(max_iter=1000,penalty=None)
X_train_selected=X_train[["petal length"]]
X_test_selected=X_test[["petal length"]]
logmodel.fit(X_train_selected, y_train)
score = logmodel.score(X_test_selected, y_test)
print(score)
```

```python
from mlxtend.feature_selection import SequentialFeatureSelector as SFS
logmodel=LogisticRegression(max_iter=1000,penalty=None)
sfs = SFS(logmodel, 
           k_features=4, 
           forward=True, 
           floating=False,
           verbose=2,
           scoring='accuracy',
           cv=5)
sfs = sfs.fit(X_train, y_train)
sfs.subsets_
```

```python
pd.DataFrame.from_dict(sfs.get_metric_dict()).T
```

```python
logmodel=LogisticRegression(max_iter=1000,penalty=None)
X_train_selected=X_train[["sepal width", "petal length", "petal width"]]
X_test_selected=X_test[["sepal width", "petal length", "petal width"]]
logmodel.fit(X_train_selected, y_train)
score = logmodel.score(X_test_selected, y_test)
print(score)
```

```python
from mlxtend.feature_selection import SequentialFeatureSelector as SFS
logmodel=LogisticRegression(max_iter=1000,penalty=None)
sbs = SFS(logmodel, 
           k_features=1, 
           forward=False, 
           floating=False, 
           verbose=2,
           scoring='accuracy',
           cv=5)
sbs = sbs.fit(X_train, y_train)
sbs.subsets_
```

```python
pd.DataFrame.from_dict(sbs.get_metric_dict()).T
```

```python
#read in the data using pandas
df = pd.read_table('diabetesdata.txt')
#check data has been read in properly
df
```

```python
X = df.drop('Diabetes',axis=1)
y = df['Diabetes']
```

```python
from sklearn.model_selection import train_test_split
# Split data into training and test sets
X_train, X_test , y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1, stratify=y)
```

```python
# Create Logistic Regression classifier
logmodel = LogisticRegression(max_iter=1000,penalty=None)
from mlxtend.feature_selection import ExhaustiveFeatureSelector as EFS
efs = EFS(logmodel, 
           min_features=1,
           max_features=8,
           scoring='accuracy',
           print_progress=True,
           cv=5)
#feature_names = ('Preg', 'Glucose', 'BP', 'SkinThick', 'Insul','BMI','DPF','Age')
efs = efs.fit(X_train, y_train)

print('Best accuracy score: %.2f' % efs.best_score_)
print('Best subset (indices):', efs.best_idx_)
print('Best subset (corresponding names):', efs.best_feature_names_)
```

```python
pd.DataFrame.from_dict(efs.get_metric_dict()).T
```

```python
pd.DataFrame.from_dict(efs.get_metric_dict()).T.sort_values('avg_score',ascending=False)
```
