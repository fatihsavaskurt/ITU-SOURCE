## Variable Types

Numerical variables can be divided into two:


* **Discrete**
* **Continuous**

As the name suggests, discrete variables take only a limited number of categories, while continuous variables can potentially take an unlimited number of values. This distinction between variable types is quite important:

* **Firstly**, in classification problem, we use discrete variables whereas in regression problem, we analyze continuous variables. 

* **Secondly**, analyzing these two types of variables requires us to apply different techniques when we begin exploring the data.



Qualitative data is obtained by dividing the data into non-numerical groups. For example, data related to a gender of a person can be converted into categorical data by 1 for man and 2 for woman. The result of mathematical operation in this variable type is not meaningful. The types of qualitative variable are:

* **Nominal**
* **Ordinal**
* **Binary**


### Categorical Variables

`Categorical variables` take limited values and represented by categories or groups. By definition, a categorical variable can only take a limited number of different values. So, categorical variables are not true measurements or counts but take on values which can only be categorized or grouped.

There are three types of categorical variables:
- Nominal variables
- Ordinal variables
- Binary variables

#### Nominal Variables

In summary, nominal variables are the variables that divide the properties  of the units that cannot be directly measured by numbers. Different categories are represented, for instance, by 0, 1, 2. However, these codes have no numerical meaning. For example, marital status; it is a four-category variable that can be classified as married, single, widowed, divorced, and this variable can be coded as “married: 0”, “single: 1”, “widow: 2” “divorced: 3”, but these numerical codes are numeric. It does not represent a size nor can be ordered.

#### Ordinal Variables

These are the variables in which their categories are listed sequentially. It is frequently used in obtaining opinions about expressions, especially in the survey studies. 

So, ordinal variables indicate the order. For example, runners are orders like first place,  second place, and 3 for third place, etc. However, this does not provide any information about the distance between places. We know that the 1st runner is faster at the 2nd runner, but the difference in their time can be minutes or milliseconds. Also, the difference between the time interval between the 1st and the 2nd runner is probably not the same as the time difference between the 2nd and the 3rd.



Another example can be about education level and it can be measured by primary, high school, university and graduate categories, it is clearly seen that there is a ranking among the categories. In short, if you can order but have no idea about the difference of the variables then it is `ordinal variable`.

## Binary Variables

`Binary variables` take two values and this makes it a special type of categorical variable. These variables are usually categorical in nature and are called binary or dichotomous variables. For instance, `male` and `female`, `yes` and `no`, and `rich` and `poor` are examples of binary variables.

## Changing the Variable Types

Depending on the case, we might need to convert a continuous variable or a categorical variable into a discrete variable.

For the sake of practice, let's discretize the `SalePrice` as high, medium, and low. Based on the summary statistics:

* Sale price>USD200.000 is labeled as 1
* USD200.000>Sale price<USD150.000 is labeled as 2
* Sale price<USD150.000 is labeled as 3

```python
house_df['SalePrice'].describe()
```

```python
def discretization(SalePrice):
    if SalePrice >= 200000:
        return 1
    elif (SalePrice > 150000) and (SalePrice < 200000):
        return 2
    else:
        return 3
```

```python
house_df['sale_discrete'] = house_df['SalePrice'].apply(discretization)
display(house_df.head())
```

```python
house_df.groupby("sale_discrete")["SalePrice"].count()
```

Or we can categorize it as follows:

* Sale price>USD200.000 is labeled as 1
* USD200.000>Sale price<USD150.000 is labeled as 2
* Sale price<USD150.000 is labeled as 3

```python
def categorization(SalePrice):
    if SalePrice >= 200000:
        return 'high'
    elif (SalePrice > 150000) and (SalePrice < 200000):
        return 'medium'
    else:
        return 'low'
```

```python
house_df['sale_cat'] = house_df['SalePrice'].apply(categorization)
display(house_df.head())
```

```python
house_df.groupby("sale_discrete")["SalePrice"].count()
```
