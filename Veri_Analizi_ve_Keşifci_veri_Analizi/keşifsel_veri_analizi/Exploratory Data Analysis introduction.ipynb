### Get frequency of categorical variables

## Visualization

```python
plt.figure(figsize=(10,6))
sns.countplot(x="Neighborhood", data=house_df)
plt.xticks(rotation=90)
plt.show()
```

## sorting the bar chart would be better

```python
plt.figure(figsize=(10,6))
sns.countplot(x="Neighborhood", data=house_df,order = house_df['Neighborhood'].value_counts().index)
plt.xticks(rotation=90)
plt.show()
```
