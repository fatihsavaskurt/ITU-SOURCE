### Handling with Missing Variable

Here are some potential reasons why we have missing values:

* Duplicate entries for the same credential during data collection.
* Errors occur during data entry process such as having some inappropriate values in the dataset.
* There may be some structural problems in the data. For example, some integer values can be stored as text in the database, or the formats of the date fields may differ from what we expected.

### Dealing with missing values 

In general, we have three options for dealing with missing values:

* Ignoring rows with missing values: Ignoring and deleting missing values is always an option. But before making this decision, one needs to think it through in that deleting may mean losing information. Because each observation in a dataset provides some information, and this information can be invaluable in many cases. In the case where we have large sample size, deleting missing values can be wise approach to follow. However, in this case of small sample size, we need to find another way to deal with missing value.



* Filling missing values with some proper values: This approach is probably the most common method that many data scientists follow when they encounter variables with missing values. When you have a small sample size or you think that there is something wrong in data entring process or variable shows, to a large extent, time invariant property, we impute the missing values. However, there is no one-size-fits all procedure imputing. As always, data scientist should consider her data characteristics before making any decision.  For instance, it is plausible to fill in the missing values with mean or median depending on the form of the distribution that the data has. For imbalanced data, the median may be more appropriate, while for symmetrical and more normally distributed data, the mean could be a better choice.

Here are some techniques that can be applied for imputation:

* Filling with 0 
* Filling with randomly selected value
* Filling with central tendency measures, i.e., mean, median or mode
* Filling with interpolation


* Run predictive models that impute the missing data. Prediction results provide us important insights about the missing values and using this information, imputing missing values can be very effective and increase the accuracy.

### Filling Missing Value with Imputation

First we try to fill missing value with mean value of the same variable. To do that, we pick `LotFrontage` variable, which has 259 missing values.

```python
house_df['LotFrontage'].isnull().sum()
```

```python
house_df['LotFrontage'].describe()
```

To fill the missing value with mean, we use `fillna()` function in Python and to observe the changes after imputing, another variable is introduced named `LotFrontage_imputed`.

```python
house_df["LotFrontage_imputed"]=house_df["LotFrontage"].fillna(house_df["LotFrontage"].mean())
```

Well, it turns out that we get rid of missing values

```python
house_df["LotFrontage_imputed"].isnull().sum()
```

```python
house_df["LotFrontage_imputed"].describe()
```

In the case of `LotFrontage`, there is no big change in mean or any other statistics, we imply that there is no big harm in applying imputing to this variable. But is this the case always? No. let's examine another variable.

```python
house_df["GarageYrBlt"].describe()
```

###  Filling Missing Values in Categorical Variable


Filling categorical variables is a bit more complicated than filling continuous variables. This is because we cannot use values that measure something like central tendency, as we do with continuous variables. Instead, we can follow two strategies:




1. We can create a new category for missing values such as other or unknown. Sometimes a categorical feature’s missingness itself is valuable information that should be explicitly encoded. If so, most-common-class imputing would cause this information to be lost. Instead, just replace those values with a value like “Unknown” or “Missing.”



2. We can try to find a suitable category among the existing categories and replace it with missing values. But how? One way is to use the most common category.

Let's check the `GarageType` variable. It seem like `Attchd` dominates the rest of garage type categories. So, it may be a wise choice to fill the missing value with the most repeated observations in the variable.

```python
house_df.GarageType.value_counts()
```

```python
house_df.GarageType.isnull().sum()
```
